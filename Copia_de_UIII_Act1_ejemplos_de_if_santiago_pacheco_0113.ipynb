### Ejemplo de if, if else, if elif

ejemplos de if

```python

# 1. Estructura if simple con operadores relacionales
# Esta estructura ejecuta un bloque de código únicamente si la condición se cumple
# Ejemplo: Verificar si un número está en un rango específico
numero = 15

if 10 <= numero <= 20:
    print(f"El número {numero} está entre 10 y 20.")
```

```python
# 2. Estructura if-else con operadores aritméticos y relacionales
# Se utiliza cuando existen dos alternativas posibles; si la condición no se cumple, se ejecuta el bloque dentro del else
# Ejemplo: Determinar si un número es par o impar
numero = 20

if numero % 2 == 0:
    print("El número es par.")
else:
    print("El número es impar.")
```

```python
# 2. Estructura if-else con operadores aritméticos y relacionales
# Se utiliza cuando existen dos alternativas posibles; si la condición no se cumple, se ejecuta el bloque dentro del else
# Ejemplo: Determinar si un número es par o impar
numero = 21

if numero % 2 == 0:
    print("El número es par.")
else:
    print("El número es impar.")
```

```python
# 3. Estructura if-elif-else con operadores lógicos y relacionales
# Permite manejar múltiples casos de forma jerárquica. Se evalúa la primera condición; si es falsa, pasa al elif y así sucesivamente
# Ejemplo: Clasificar un año como bisiesto o no
anio = 2024

if (anio % 4 == 0 and anio % 100 != 0) or (anio % 400 == 0):
    print(f"{anio} es un año bisiesto.")
else:
    print(f"{anio} no es un año bisiesto.")
```

```python
# 4. Ejemplo de descuento con if-elif-else y operadores relacionales
# Este ejemplo es ideal para ver cómo se estructuran rangos de decisión
# Ejemplo: Calcular descuento por cantidad de artículos
cantidad = 12

if cantidad >= 10:
    descuento = 0.2  # 20% de descuento
elif cantidad >= 5:
    descuento = 0.1  # 10% de descuento
else:
    descuento = 0    # Sin descuento

print(f"Tu descuento es del {descuento * 100}%.")
```
